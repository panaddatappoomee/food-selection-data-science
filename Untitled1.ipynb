```python
import random
import sqlite3
from datetime import datetime, timedelta
import pandas as pd
import matplotlib.pyplot as plt
class Book:
    def __init__(self, book_id, title, author, category_code, category_name, status=" าง"):
        self.book_id = book_id
        self.title = title
        self.author = author
        self.category_code = category_code
        self.category_name = category_name
        self.is_available = (status == " าง")
        self.loan_history = [] # New: to track all loans for this book

    def borrow(self):
        """เปลี่ยนสถานะเป็นถูกยืม"""
        self.is_available = False

    def return_book(self):
        """เปลี่ยนสถานะกลับเป็นว่าง"""
        self.is_available = True

    def add_loan(self, loan): # New method to add loan to history
        self.loan_history.append(loan)

    def is_available_at(self, check_date): # New method to check availability at a specific date
        for loan in self.loan_history:
            # A book is NOT available if there's an overlapping loan
            if loan.borrow_date <= check_date and (loan.return_date is None or loan.return_date >= check_date):
                return False # Book is currently on loan or was on loan at check_date
        return True # No overlapping loan found, so it's available

class Member:
    def __init__(self, member_id, name, member_type, loan_period_days):
        self.member_id = member_id
        self.name = name
        self.member_type = member_type
        self.loan_period_days = loan_period_days
class BookLoan:
    FINE_PER_DAY = 5
    def __init__(self, loan_id, book: Book, member: Member, borrow_date):
        self.loan_id = loan_id
        self.book = book
        self.member = member
        self.borrow_date = borrow_date
        self.due_date = borrow_date + timedelta(days=member.loan_period_days)
        self.return_date = None


    def calculate_late_fee(self):
        """คืนจำนวนวันที่ยืมได้ ตามประเภทสมาชิก"""
        if self.return_date is not None and self.return_date > self.due_date:
            late_days = (self.return_date - self.due_date).days
            return late_days * self.FINE_PER_DAY
        return 0
    def mark_returned(self, return_date):
        """บันทึกการคืนว่าเมื่อไหร่"""
        self.return_date = return_date
        self.book.return_book()
```

```python
class BookLoan:

    def __init__(self, loan_id, book, member, borrow_date):
        self.loan_id = loan_id
        self.book = book
        self.member = member
        self.borrow_date = borrow_date
        self.due_date = borrow_date + timedelta(days=member.loan_period_days)
        self.return_date = None
        self.status = "borrowed"

    def mark_returned(self, return_date):
        self.return_date = return_date
        if return_date <= self.due_date:
            self.status = "on_time"
        else:
            self.status = "late"

    def calculate_late_fee(self, rate_per_day=5):
        """คำนวณค่าปรับกรณีคืนสาย (คิดวันละ rate_per_day บาท เช่น 5 บาท/วัน)"""
        if self.return_date and self.return_date > self.due_date:
            late_days = (self.return_date - self.due_date).days
            return late_days * rate_per_day
        return 0
```

```python
from datetime import timedelta
import random
import pandas as pd


def load_books(csv_path):
    """โหลดหนังสือทั้งหมดจากไฟล์ CSV แล้วคืนค่าเป็น list ของ Book object"""
    df = pd.read_csv('/content/books.csv')
    return [
        Book(
            row.book_id,
            row.title,
            row.author,
            row.category_code,
            row.category_name,
            row.status,
        )
        for row in df.itertuples(index=False)
    ]


def load_members(csv_path):
    """โหลดสมาชิกทั้งหมดจากไฟล์ CSV แล้วคืนค่าเป็น list ของ Member object"""
    df = pd.read_csv('/content/members.csv')
    return [
        Member(row.member_id, row.name, row.member_type, row.loan_period_days)
        for row in df.itertuples(index=False)
    ]


def random_borrow_date(start_date, days_range=300):
    """สุ่มวันที่ยืม ภายในช่วง days_range วันจาก start_date (default = 300 วัน)"""
    offset = random.randint(0, days_range)
    return start_date + timedelta(days=offset)


def decide_return_outcome(
    never_return_probability=0.20, late_probability=0.25
):
    """สุ่มผลลัพธ์การคืนหนังสือ: 'on_time' / 'late' / 'not_returned'"""
    r = random.random()
    if r < never_return_probability:
        return "not_returned"
    elif r < never_return_probability + late_probability:
        return "late"
    else:
        return "on_time"


def calculate_return_date(due_date, outcome):
    """คำนวณวันที่คืนจริงจากผลลัพธ์ที่สุ่มได้ (คืนค่า None ถ้ายังไม่คืน)"""
    if outcome == "not_returned":
        return None
    elif outcome == "late":
        return due_date + timedelta(days=random.randint(1, 10))
    else:
        return due_date - timedelta(days=random.randint(0, 5))
```

```python
random.seed(1)
```

```python
books = load_books("books.csv")
members = load_members("members.csv")
```

```python
loans = []
start_date = datetime(2025, 1, 1)
```

```python
def simulate_library_visit(loan_id, member, books, start_date):
    # ดึงรายชื่อหนังสือที่ยังว่างอยู่
    available_books = [b for b in books if b.is_available]

    # ถ้าหนังสือถูกยืมไปจนหมด ให้ Reset สถานะหนังสือทั้งหมดเป็นว่างเพื่อจำลองต่อ
    if not available_books:
        available_books = books
        for b in available_books:
            b.return_book()

    # สุ่มเลือกหนังสือ
    book = random.choice(available_books)

    # กำหนดวันที่ยืม
    borrow_date = random_borrow_date(start_date)

    # สร้าง Object รายการยืม และเปลี่ยนสถานะหนังสือเป็นถูกยืม
    loan = BookLoan(loan_id, book, member, borrow_date)
    book.borrow()

    # สุ่มผลลัพธ์การคืน และคำนวณวันที่คืน
    outcome = decide_return_outcome()
    return_date = calculate_return_date(loan.due_date, outcome)

    # บันทึกวันที่คืน (ถ้ามีการคืน)
    if return_date is not None:
        loan.mark_returned(return_date)

    return loan


# รายชื่อสมาชิกที่เดินเข้ามาทีละคน
walk_in_members = [
    Member(member_id="PS0001", name="รติ", member_type="นักศึกษา", loan_period_days=14),
    Member(member_id="PS0002", name="กุ๊กกิ๊ก", member_type="นักศึกษา", loan_period_days=14),
    Member(member_id="PT0001", name="อร", member_type="อาจารย์", loan_period_days=30),
]

completed_loans = []
```

```python
# วนลูปเพื่อจำลองบริการทีละคน
for i, member in enumerate(walk_in_members, start=1):
    loan = simulate_library_visit(
        loan_id=i, member=member, books=books, start_date=start_date
    )
    completed_loans.append(loan)

print("=" * 60)
print(
    f"🏁 จบรอบสาธิต — มีสมาชิกเข้ามายืมหนังสือทั้งหมด {len(completed_loans)} คน"
)
```

```python
for i in range(1, 351):
    # ดึงรายชื่อหนังสือที่ยังว่างอยู่
    available_books = [b for b in books if b.is_available]

    # ถ้าหนังสือถูกยืมไปจนหมด ให้ Reset สถานะหนังสือทั้งหมดเป็นว่างเพื่อจำลองต่อ
    if not available_books:
        available_books = books
        for b in available_books:
            b.return_book()

    # สุ่มเลือกหนังสือ และ สมาชิก
    book = random.choice(available_books)
    member = random.choice(members)

    # กำหนดวันที่ยืม และสร้าง ID การยืม เช่น L0001, L0002
    borrow_date = random_borrow_date(start_date)
    loan_id = f"L{i:04d}"

    # สร้าง Object รายการยืม และเปลี่ยนสถานะหนังสือเป็นถูกยืม
    loan = BookLoan(loan_id, book, member, borrow_date)
    book.borrow()

    # สุ่มผลลัพธ์การคืน และคำนวณวันที่คืน
    outcome = decide_return_outcome()
    return_date = calculate_return_date(loan.due_date, outcome)

    # บันทึกวันที่คืน (ถ้ามีการคืน)
    if return_date is not None:
        loan.mark_returned(return_date)

    loans.append(loan)


print(f"จำลองการยืมสำเร็จแล้วทั้งหมด {len(loans)} รายการ")
```

```python
import pandas as pd

# สร้าง List ของ Dictionary จาก Object ทั้งหมด
records = [
    {
        "loan_id": loan.loan_id,
        "book_id": loan.book.book_id,
        "member_id": loan.member.member_id,
        "borrow_date": loan.borrow_date.strftime("%Y-%m-%d"),
        "due_date": loan.due_date.strftime("%Y-%m-%d"),
        "return_date": (
            loan.return_date.strftime("%Y-%m-%d") if loan.return_date else None
        ),
        "late_fee": loan.calculate_late_fee(),  # เรียก method ของ object เพื่อคำนวณค่าปรับ
    }
    for loan in loans  # วนทุก object ใน list
]

# แปลงเป็น DataFrame และบันทึกเป็นไฟล์ CSV
df_loans = pd.DataFrame(records)
df_loans.to_csv("loans.csv", index=False, encoding="utf-8-sig")

print("บันทึกไฟล์ CSV สำเร็จ ขนาด:", df_loans.shape)
print(df_loans)
```

```python
# Join df_loans กับ df_books แล้วนับจำนวนการยืมตามหมวดหมู่
top_categories = (
    df_loans.merge(df_books, on="book_id")
    .groupby("category_name")["loan_id"]
    .count()
    .sort_values(ascending=False)
)
print(top_categories.head(3))
```

```python
# สร้าง DataFrame สำหรับ Members
member_records = [
    {
        "member_id": member.member_id,
        "name": member.name,
        "member_type": member.member_type,
        "loan_period_days": member.loan_period_days,
    }
    for member in members
]
df_members = pd.DataFrame(member_records)

# Group ตามประเภทสมาชิกแล้วรวมค่าปรับ late_fee
fee_by_member_type = (
    df_loans.merge(df_members, on="member_id")
    .groupby("member_type")["late_fee"]
    .sum()
    .reset_index()
)
print(fee_by_member_type)
```
